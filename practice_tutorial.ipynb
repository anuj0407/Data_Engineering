### StructType() Schema

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python

my_struct_schema=StructType([
    StructField('Item_Identifier',StringType(),True),
    StructField('Item_WeighT string',StringType(),True),
 StructField('Item_Fat_Content string',StringType(),True),
 StructField('Item_Visibility double',StringType(),True),
 StructField('Item_Type string',StringType(),True),
 StructField('Item_MRP double',StringType(),True),
 StructField('Outlet_Identifier string',StringType(),True),
 StructField('Outlet_Establishment_Year integer',StringType(),True),
 StructField('Outlet_Size string',StringType(),True),
 StructField('Outlet_Location_Type string',StringType(),True),
 StructField('Outlet_Type string',StringType(),True),
 StructField('Item_Outlet_Sales double',StringType(),True)
])
```

```python

df=spark.read.format('csv')\
    .schema(my_struct_schema)\
        .option('header',True)\
            .load('BigMart Sales.csv')
```

```python
df.printSchema()
```

### SELECT

```python
df.show()
```

#### Normal SELECT

```python
df.select('Item_Identifier','Item_Weight','Item_Fat_Content').show()
```

#### SELECT using col

```python

df.select(col('Item_Identifier'),col('Item_Weight'),col('Item_Fat_Content')).show()
```

### ALIAS

```python

df.select(col('Item_Identifier').alias('Item_ID')).show()
```

### FILTER/WHERE

scenario-1

```python
df.filter(col('Item_Fat_Content')=='Regular').show()
```

scenario-2

```python
df.filter( (col('Item_Type')=='Soft Drinks') & ( col('Item_Weight')<10) ).show()
```

scenario-3

```python
df.filter((col('Outlet_Size').isNull()) & (col('Outlet_location_Type').isin('Tier 1','Tier 2'))).show()
```

### WithColumnRenamed

```python
df.withColumnRenamed('Item_Weight','Item_Wt').show()
```

### WithColumn

scenario-1

```python

df = df.withColumn('flag',lit('new'))
```

```python
df.show()
```

scenario-2

```python

df=df.withColumn('multiply',col('Item_Weight')*col('Item_MRP'))
```

```python
df.show()
```

```python

df.withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),'Regular','Reg'))\
    .withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),'Low Fat','Lf')).show()
```

### Type Casting

```python

df=df.withColumn('Item_Weight',col('Item_Weight').cast(StringType()))
```

```python
df.printSchema()
```

### UNION AND UNION BY NAME

```python
data1=[('1','kad'),
       ('2','sid')]
schema1='id STRING , name STRING'

df1=spark.createDataFrame(data1,schema1)

data2=[('3','rahul'),
       ('4','jas')]
schema2='id STRING , name STRING'

df2=spark.createDataFrame(data2,schema2)
```

```python

df2.show()
```

```python
df1.show()
```

union


```python
df1.union(df2).show()
```

```python

data1 = [('kad','1',),
        ('sid','2',)]
schema1 = 'name STRING, id STRING' 

df1 = spark.createDataFrame(data1,schema1)

df1.show()
```

```python
df1.union(df2).show()
```

### Union by Name

```python
df1.unionByName(df2).show()
```

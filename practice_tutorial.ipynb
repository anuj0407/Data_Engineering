### StructType() Schema

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python

my_struct_schema=StructType([
    StructField('Item_Identifier',StringType(),True),
    StructField('Item_WeighT string',StringType(),True),
 StructField('Item_Fat_Content string',StringType(),True),
 StructField('Item_Visibility double',StringType(),True),
 StructField('Item_Type string',StringType(),True),
 StructField('Item_MRP double',StringType(),True),
 StructField('Outlet_Identifier string',StringType(),True),
 StructField('Outlet_Establishment_Year integer',StringType(),True),
 StructField('Outlet_Size string',StringType(),True),
 StructField('Outlet_Location_Type string',StringType(),True),
 StructField('Outlet_Type string',StringType(),True),
 StructField('Item_Outlet_Sales double',StringType(),True)
])
```

```python

df=spark.read.format('csv')\
    .schema(my_struct_schema)\
        .option('header',True)\
            .load('BigMart Sales.csv')
```

```python
df.printSchema()
```

### JOINS

```python
dataj1 = [('1','gaur','d01'),
          ('2','kit','d02'),
          ('3','sam','d03'),
          ('4','tim','d03'),
          ('5','aman','d05'),
          ('6','nad','d06')] 

schemaj1 = 'emp_id STRING, emp_name STRING, dept_id STRING' 

df1 = spark.createDataFrame(dataj1,schemaj1)

dataj2 = [('d01','HR'),
          ('d02','Marketing'),
          ('d03','Accounts'),
          ('d04','IT'),
          ('d05','Finance')]

schemaj2 = 'dept_id STRING, department STRING'

df2 = spark.createDataFrame(dataj2,schemaj2)
```

```python
df1.show()
```

```python
df2.show()
```

#### Inner Join

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'inner').show()
```

#### Left Join

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'left').show()
```

##### Right Join

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'right').show()
```

##### Anti Join

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'anti').show()
```

##### ROW NUMBER()

```python
from pyspark.sql.window import Window
```

```python
df.withColumn('rowCol',row_number().over(Window.orderBy('Item_MRP'))).show()
```

##### RANK VS DENSE RANK

```python
df.withColumn('rowCol',row_number().over(Window.orderBy('Item_Identifier'))).show()
```

```python
df.withColumn('rank',rank().over(Window.orderBy(col('Item_Identifier').desc())))\
        .withColumn('denseRank',dense_rank().over(Window.orderBy(col('Item_Identifier').desc()))).show()
```

##### CUMULATIVE SUM

```python
df.withColumn('CumSum',sum('Item_MRP').over(Window.orderBy('Item_Type'))).show()
```

```python

df.withColumn('cumsum',sum('Item_MRP').over(Window.orderBy('Item_Type').rowsBetween(Window.unboundedPreceding,Window.currentRow))).show()
```

```python
df.withColumn('totalsum',sum('Item_MRP').over(Window.orderBy('Item_Type').rowsBetween(Window.unboundedPreceding,Window.unboundedFollowing))).show()
```
